# Setting up <strong>your</strong> ADF run - optional

If you have a case or cases you want to run the ADF with, this section will help with a quick setup. This will include basic information about the configuration yaml files and how to run the ADF for your cases!

Instead, if you would like a guided demo of how to use the ADF with sample data, please skip to the <a href="https://justin-richling.github.io/ADF-Tutorial/notebooks/basic_examples/readme.html">GUIDED EXAMPLES</a> section.


0. Make sure your ADF clone is the most up to date version

* Go to where you installed the ADF, make sure you're on the correct branch and run `git pull origin <branch>`. Most cases you'll just use the main branch, but there are others (as of 05/2026) for:

<ul>
<li>Native Grid (gridding to lat/lon or unstructured plotting via `uxarray`) <code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">native-grid</span></code></li>
<li>Multi-Case <code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">multi-case-diags</span></code></li>
<li>WACCM diagnostics <code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">waccm-diags</span></code></li>
</ul>

<p>1. make a copy of the config_cam_baseline_example.yaml file, feel free to name it something that makes sense to you. The yaml files are in the root folder of where you installed the ADF.</p>

<code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">&emsp;config_cam_baseline_example.yaml</span></code>

</p>2. open that copied file and the main sections you'll want to change are:
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">user</span></code> (optional) - use your NCAR username as a variable in the yaml file
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">diag_basic_info</span></code> - this is where you'll set some main run variables, the ones I usually change are:
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">compare_obs</span></code> - set true/false depending which run you want
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_regrid_loc</span></code> - where the regridded files for the cases in this ADF run will be saved, you can change or use the default location, just take note
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_diag_loc</span></code> - where the diagnostic plots will be saved, again, you can change or use default
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">diag_cam_climo</span></code> - for the experiment case (test), the ones I change are:
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_case_name</span></code> - just the name of the case run
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_hist_loc</span></code> - where the h# history files live
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">case_nickname</span></code> - anything that would be easier than the case name if you want, it will default to the case name if left blank
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">start/end_years</span></code> - climo years desired
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_climo_loc</span></code> - where you want climo files saved
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">cam_ts_loc</span></code>  - where you want time series files saved
<p><code class="docutils literal notranslate"><span class="pre" style="color: #076E85;">diag_cam_baseline_climo</span></code> - same exact variables as the diag_cam_climo above, just set for the control (baseline) case. If using Obs, just ignore this section</p>

Then at the end of the config file are the plots and variables, set these as you like!</p>

</p>3. To run the ADF (from the ADF root path) in a terminal just run: ./run_adf_diag your_new_yaml_file.yaml</p>
